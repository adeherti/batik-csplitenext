# Gambar R1-12: contoh salah klasifikasi (tanpa GPU, ~2 menit)

Membuat Gambar contoh salah klasifikasi r = 0.16 dari prediksi pengulangan CV (fold seed 2026). Daftar citra sudah dipilih otomatis: untuk empat pasangan kelas yang paling sering tertukar, diambil citra salah klasifikasi dengan keyakinan tertinggi, lalu dibandingkan dengan citra representatif kelas prediksi dan kelas sebenarnya (citra yang diklasifikasi benar dengan keyakinan tertinggi).

**Sebelum menjalankan:** Add Input `batik-clean-v1`. Accelerator **None** (tidak perlu GPU). Lalu Save & Run All. Unduh `fig_misclassified.png` dan `fig_misclassified.pdf` dari tab Output.

```python
import os, json
import matplotlib.pyplot as plt
from PIL import Image

SEL = json.loads(r'''[
 {
  "mis": "val/Solo_Parang/Solo_Parang_val_00003.jpg",
  "true": "Solo_Parang",
  "pred": "Yogyakarta_Parang",
  "conf": 0.96,
  "ref_pred": "train/Yogyakarta_Parang/Yogyakarta_Parang_train_00004.jpg",
  "ref_true": "train/Solo_Parang/Solo_Parang_train_00027.jpg"
 },
 {
  "mis": "train/Madura_Mataketeran/Madura_Mataketeran_train_00000.jpg",
  "true": "Madura_Mataketeran",
  "pred": "Sulawesiselatan_Lontara",
  "conf": 0.89,
  "ref_pred": "train/Sulawesiselatan_Lontara/Sulawesiselatan_Lontara_train_00010.jpg",
  "ref_true": "test/Madura_Mataketeran/Madura_Mataketeran_test_00000.jpg"
 },
 {
  "mis": "train/Aceh_Pintu_Aceh/Aceh_Pintu_Aceh_train_00000.jpg",
  "true": "Aceh_Pintu_Aceh",
  "pred": "Lampung_Gajah",
  "conf": 0.9,
  "ref_pred": "train/Lampung_Gajah/Lampung_Gajah_train_00014.jpg",
  "ref_true": "train/Aceh_Pintu_Aceh/Aceh_Pintu_Aceh_train_00002.jpg"
 },
 {
  "mis": "train/Papua_Cendrawasih/Papua_Cendrawasih_train_00018.jpg",
  "true": "Papua_Cendrawasih",
  "pred": "Papua_Tifa",
  "conf": 0.7,
  "ref_pred": "val/Papua_Tifa/Papua_Tifa_val_00002.jpg",
  "ref_true": "train/Papua_Cendrawasih/Papua_Cendrawasih_train_00003.jpg"
 }
]''')

ROOT = None
for r, d, f in os.walk("/kaggle/input"):
    if {"train", "val", "test"}.issubset(set(d)):
        ROOT = r; break
assert ROOT, "Dataset batik-clean-v1 tidak ditemukan. Add Input dulu."
print("ROOT:", ROOT)

def nice(c): return c.replace("_", " ")
cols = ["Misclassified", "Typical of predicted class", "Typical of true class"]
fig, axes = plt.subplots(len(SEL), 3, figsize=(9, 3.1 * len(SEL)))
for r, s in enumerate(SEL):
    imgs = [s["mis"], s["ref_pred"], s["ref_true"]]
    caps = [f"True: {nice(s['true'])}\nPredicted: {nice(s['pred'])} ({s['conf']:.2f})",
            nice(s["pred"]), nice(s["true"])]
    for c in range(3):
        ax = axes[r, c]
        ax.imshow(Image.open(os.path.join(ROOT, imgs[c])).convert("RGB").resize((224, 224)))
        ax.set_xticks([]); ax.set_yticks([])
        ax.set_xlabel(caps[c], fontsize=8)
        if r == 0: ax.set_title(cols[c], fontsize=9, fontweight="bold")
        if c == 0:
            for sp in ax.spines.values(): sp.set_edgecolor("#c0392b"); sp.set_linewidth(2)
plt.tight_layout()
plt.savefig("/kaggle/working/fig_misclassified.png", dpi=300)
plt.savefig("/kaggle/working/fig_misclassified.pdf")
print("selesai")
```
